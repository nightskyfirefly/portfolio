```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# Set up the WebDriver
driver = webdriver.Chrome('D:\\downloads\\chromedriver-win64\\chromedriver.exe')

# Open the page
driver.get('https://www.fragrantica.com/search/')

perfumes_data = []

# ... [initialization and setup code as before]

try:
    while len(perfumes_data) < 50:
        # Wait until the perfume elements are loaded
        WebDriverWait(driver, 30).until(
            EC.visibility_of_all_elements_located((By.CSS_SELECTOR, 'div.cell.card.fr-news-box'))
        )
        # Find all the perfume container elements
        containers = driver.find_elements(By.CSS_SELECTOR, 'div.cell.card.fr-news-box')
        for container in containers:
            # Extract the perfume name and perfumer from the elements
            perfume_name_element = container.find_element(By.CSS_SELECTOR, 'p > a')
            perfume_brand_element = container.find_element(By.CSS_SELECTOR, 'p > small')
            
            perfume_name = perfume_name_element.text.strip()
            perfume_brand = perfume_brand_element.text.strip()
            link = perfume_name_element.get_attribute('href')
            
            perfumes_data.append({"Perfume Name": perfume_name, "Perfume Brand": perfume_brand, "Link": link})
            
            if len(perfumes_data) >= 50:
                break

except Exception as e:
    print("Scraping completed or an error occurred:", e)

finally:
    # Correcting specific order issues if necessary (example shown, adapt as needed)
    # This is a placeholder for any manual swaps or sorting you might need
    # perfumes_data[2], perfumes_data[3] = perfumes_data[3], perfumes_data[2]

    df = pd.DataFrame(perfumes_data)

    if not df.empty:
        df.to_csv('top_50_perfumes.csv', index=False)
        print("CSV file has been saved.")
    else:
        print("No data was scraped.")

```
