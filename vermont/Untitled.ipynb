```python
# Re-importing necessary library since the code execution state was reset
import pandas as pd

# Define the provided data
age_group_counts = {'Gen Z': 15, 'Millennials': 34, 'Gen X': 31, 'Baby Boomers': 20}
total_claims = {'Gen Z': 85, 'Millennials': 143, 'Gen X': 145, 'Baby Boomers': 87}
total_annual_spend = {'Gen Z': 182413.8, 'Millennials': 319395.73, 'Gen X': 298082.4, 'Baby Boomers': 237205.83}

# Calculate total annual spend for all age groups
total_annual_spend_all_groups = sum(total_annual_spend.values())

# Calculate the percentage of total spend for each age group
percentage_of_total_spend = {group: (total_annual_spend[group] / total_annual_spend_all_groups) * 100 
                             for group in total_annual_spend}

# Calculate average spend per claim for each age group
average_spend_per_claim = {group: (total_annual_spend[group] / total_claims[group])
                           for group in total_claims}

# Calculate average spend per person in each age group
average_spend_per_person = {group: (total_annual_spend[group] / age_group_counts[group])
                            for group in age_group_counts}

(percentage_of_total_spend, average_spend_per_claim, average_spend_per_person)
```
